```python
model1=LogisticRegression()
model2=SVC()
model3=RandomForestClassifier()
model4=GradientBoostingClassifier(n_estimators=1000)
```

```python
def call(model):
    model.fit(xtrain,ytrain)
    model.score(xtrain,ytrain)
    model.score(xtest,ytest)
    # anothe way
    per =model.predict(xtest)
    print(accuracy_score(per,ytest))
    print(recall_score(per,ytest))
    print(f1_score(per,ytest))
    print(confusion_matrix(per,ytest))
    print(sns.heatmap(confusion_matrix(per,ytest),annot=True))
call(model1)
```
