### map

```python
df['pclass'] = df['pclass'].map({1:'one',2:'tow',3:'three'})
```

```python
df.head(10)
```

```python
plt.figure(figsize=(20,10))
df['fare'].plot()
```

```python
df.describe().round(2)
```

```python
df['fare']=((df['fare']-df['fare'].mean())/df['fare'].std()).round(2)
```

```python
df
```

```python
plt.figure(figsize=(20,10))
df['fare'].plot()
```

### binning : make the data is cuting

```python
df.head(3)
```

```python
pd.cut(df['age'],bins=[0,10,20,30,45,60,70,100]).value_counts()
```

```python
df['groupage'] =pd.cut(df['age'],bins=[0,10,20,30,45,60,70,100],
                      labels=['child','teen','adult','youth','midleaged','senior','old'])
```

```python
df.head(5)
```
