```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import preprocessing
```

```python
data = pd.read_csv("C:\\Users\\G.B\\Desktop\\osam\\data analysis\\student.csv",header=0)
#print(data)
#print(data.isnull().sum())
# clean data
data=data.drop(["test preparation course","race/ethnicity"],axis=1)
print(data.columns)

# divid data object and int
obj_data=data.select_dtypes(include=["object"])
print(obj_data)
# labeling data
lb=preprocessing.LabelEncoder()
for i in range(obj_data.shape[1]):# columns
    # take columns and columns and take value to same columns
    obj_data.iloc[:,i]=lb.fit_transform(obj_data.iloc[:,i])
#print(obj_data)

num_data=data.select_dtypes(exclude=["object"])
data = pd.concat([obj_data,num_data],axis=1)
print(data) # all value in data is number

# change all value string to number
# relashineship
c =data.corr()
print(c)
sns.heatmap(c,annot=True)
plt.show
```
