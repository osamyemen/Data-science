### conver null values in columns

```python
# way : mean , median , most_frequent , constant
data=data['Distance'].fillna(data['Distance'].median())
```

### covert all null values in data

```python
# method : mean , median , most_frequent , constant
imputer=SimpleImputer(missing_values=np.nan,strategy="mean")
clear_data=imputer.fit_transform(data)
data=pd.DataFrame(clear_data,columns=data.columns)
```

### convert text data to number data

```python
lb=preprocessing.LabelEncoder()
obj_data=data.select_dtypes(include=["object"])
#print(obj_data)
num_data=data.select_dtypes(exclude=["object"])
for i in range(0,obj_data.shape[1]):
    obj_data.iloc[:,i]=lb.fit_transform(obj_data.iloc[:,i])
data = pd.concat([obj_data,num_data],axis=1)
#data.head()
```
