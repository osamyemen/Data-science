# neural_network_regression

```python
from sklearn import preprocessing
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
```

```python
data=pd.read_csv("C:\\Users\\G.B\\Desktop\\المرحلة الاولى\\data science and machine learning\\dataset\\Iris.csv")


obj_data=data.select_dtypes(include=["object"])
lb=preprocessing.LabelEncoder()

for i in range(obj_data.shape[1]):
    obj_data.iloc[:,i]=lb.fit_transform(obj_data.iloc[:,i])
num_data=data.select_dtypes(exclude=["object"])
data = pd.concat([obj_data,num_data],axis=1)






data=data.dropna()
print(data.head())
print(data.isnull().sum())
x=data.iloc[:,:-1]
y=data.iloc[:,-1]
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.90)
```

```python
model =MLPRegressor(activation="relu",max_iter=1000)
model.fit(x_train,y_train)
print(model.score(x_train,y_train))
print(model.score(x_test,y_test))
```

# neural_network_logisticregression

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
```

```python
data=pd.read_csv("C:\\Users\\G.B\\Desktop\\المرحلة الاولى\\data science and machine learning\\dataset\\.csv")
#print(data.head())
print(data.isnull().sum())
data=data.dropna()
x=data.iloc[:,:-1]
y=data.iloc[:,-1]
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.90)
```

```python
model=MLPClassifier(hidden_layer_sizes=(64,128,32),learning_rate="constant",
                   learning_rate_init=0.002,max_iter=1000)
```

```python
model.fit(x_train,y_train)
print(model.score(x_train,y_train))
model.score(x_test,y_test)
```

```python
model.predict([[5.1,3.5,1.4,0.2]])
```
