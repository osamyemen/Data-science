```python
import sqlite3

# Create a SQL connection to our SQLite database
con = sqlite3.connect(R"C:\Users\G.B\Desktop\osam\المرحلة الثانية\data science\dataset\3\database.db")

cur = con.cursor()
cur.execute('SELECT * FROM Reviews')
s =[]
# The result of a "cursor.execute" can be iterated over by row
for row in cur:
    s.append(row)

# Be sure to close the connection
con.commit()
con.close()
```
